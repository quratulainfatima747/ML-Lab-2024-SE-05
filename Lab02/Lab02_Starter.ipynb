## Task 8 · Inflation FY2024 vs FY2025 (real data, Pakistan Economic Survey 2024-25)
`cpi` has 2 rows (FY2024, FY2025) and 10 columns (Jul to Apr), year-on-year CPI inflation in %.
- **(a)** Average inflation of each year.
- **(b)** FY2024 months with inflation above 25 %.
- **(c)** Drop in inflation for each month (FY2024 − FY2025).
- **(d)** The month with the biggest drop.
- **(e)** Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with `np.where`.

**Interpret:** in 1–2 lines, what happened to prices in Pakistan between the two years?

```python
monsoon_total = monsoon.sum(axis=1)

wettest = np.argmax(monsoon_total)

print("Wettest City:", np.array([
    "Lahore", "Karachi", "Islamabad",
    "Peshawar", "Quetta"
])[wettest])

print("Monsoon Rainfall:", monsoon_total)

share = (monsoon_total / rain.sum(axis=1)) * 100

print("Monsoon Share Percentage:", share)
average = cpi.mean(axis=1)

print("Average Inflation:", average)
months = np.array([
    "Jul", "Aug", "Sep", "Oct", "Nov",
    "Dec", "Jan", "Feb", "Mar", "Apr"
])

high = months[cpi[0] > 25]

print("Months Above 25%:", high)
drop = cpi[0] - cpi[1]

print("Monthly Inflation Drop:", drop)
index = np.argmax(drop)

print("Biggest Drop Month:", months[index])
categories = np.where(
    cpi[1] < 5,
    "Low",
    np.where(cpi[1] < 10, "Moderate", "High")
)

print("Inflation Categories:", categories)

```

```python
Interpretation:

Inflation decreased significantly in FY25 compared with FY24. 
This means prices were increasing at a slower rate than before. 
However, it does not mean that the prices of goods and services decreased.
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
new_marks = np.clip(marks + 5, 0, 100)

print("New Marks:", new_marks)
passed = new_marks >= 50

percentage = passed.mean() * 100

print("Passing Percentage:", percentage)
grace_passed = (marks < 50) & (new_marks >= 50)

print("Students Passed Due to Grace:", np.where(grace_passed)[0])
z_scores = (marks - marks.mean()) / marks.std()

print("Z-Scores:", z_scores)
```
