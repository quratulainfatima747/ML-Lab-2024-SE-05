Task_06

What is the main difference between a tuple and a list?

The main difference between a list and a tuple in Python is that a list is mutable, whereas a tuple is immutable.

List: Its elements can be changed, added, or removed after creation. Lists use square brackets [].
Tuple: Its elements cannot be changed, added, or removed after creation. Tuples use parentheses ().
Example:
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

Task_08

Function: return True if the word "dog" is in a string (case-insensitive).

```python
def find_dog(text):
    return "dog" in text.lower().split()

print(find_dog("Is there a dog here?"))
print(find_dog("I have a DOG"))
print(find_dog("I have a cat"))
```

Task_09

Function: count how many times "dog" occurs in a string (case-insensitive)

```python
def count_dog(text):
    return text.lower().split().count("dog")

print(count_dog("dog Dog cat dog"))
```

Task_10

caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed = speed - 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"


print(caught_speeding(60, False))
print(caught_speeding(65, True))
print(caught_speeding(75, False))
print(caught_speeding(85, False))
```
