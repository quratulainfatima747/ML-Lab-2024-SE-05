Task#8

Statement:

Function: return True if the word "dog" is in a string (case-insensitive).

```python
def find_dog(text):
    return 'dog' in text.lower()

print(find_dog("Is there a Dog here?"))  
print(find_dog("Is there a cat here?"))  
print(find_dog("Dog is here"))
```

Task#9

Statement:

Function: count how many times "dog" occurs in a string (case-insensitive)

```python
def count_dog(text):
    return text.lower().count('dog')

print(count_dog("This dog runs faster than the other DOG"))
```

Task#10

Statement:

caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    limit_boost = 5 if is_birthday else 0
    if speed <= 60 + limit_boost:
        return "No Ticket"
    elif speed <= 80 + limit_boost:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(caught_speeding(60, False))  
print(caught_speeding(65, True))  
print(caught_speeding(70, False)) 
print(caught_speeding(85, True))   
print(caught_speeding(90, True))   
```
