## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
print("T3a: Missing Values Before Cleaning")
print(df.isnull().sum())
before = len(df)

df = df.dropna(subset=["Brand", "Model"])

removed = before - len(df)

df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")

print("T3a: Rows Removed:", removed)
print("T3a: Remaining Rows:", len(df))
df["Registered City"] = (
    df["Registered City"]
    .str.strip()
    .str.title()
)

df["Price"] = (
    df["Price"]
    .astype(str)
    .str.replace("Rs", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)

print("T3b: Formatting Fixed")
display(df[["Registered City", "Price"]].head())
df["Price"] = pd.to_numeric(df["Price"]).astype("int64")

df["Year"] = pd.to_numeric(
    df["Year"],
    errors="coerce"
).astype("Int64")

print("T3c: Updated Data Types")
print(df.dtypes)
print("T3d: Cities Before Correction:")
print(df["Registered City"].nunique())

df["Registered City"] = df["Registered City"].replace({
    "Sukkar": "Sukkur"
})

df["Model"] = df["Model"].replace({
    "Corrolla Altis": "Corolla Altis"
})

print("T3d: Cities After Correction:")
print(df["Registered City"].nunique())
print("T3e: Duplicate Rows Before Cleaning:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("T3e: Rows After Removing Duplicates:")
print(len(df))
invalid_km = (
    (df["KMs Driven"] < 100) |
    (df["KMs Driven"] > 1_000_000) |
    (df["KMs Driven"].isin([12345, 123456]))
)

df.loc[invalid_km, "KMs Driven"] = np.nan

fake_prices = [12345678, 87654321]

invalid_price = (
    df["Price"].isin(fake_prices) |
    (
        df["Brand"].isin(["Suzuki", "Daihatsu"]) &
        (df["Price"] > 5_000_000)
    )
)

df = df[~invalid_price]

print("T3f: Invalid Values Cleaned")
print("Remaining Rows:", len(df))
print("T3g: Final Dataset Shape")
print(df.shape)
```

```python
In this task, we cleaned the dataset by handling missing values, correcting formatting problems, converting data types,fixing inconsistent categories,
removing duplicate records, and handling invalid values.
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
df["Price_lakh"] = df["Price"] / 100000

print("T4a: Price in Lakhs")
display(df[["Price", "Price_lakh"]].head())
df["Age"] = 2020 - df["Year"]

print("T4b: Car Age")
display(df[["Year", "Age"]].head())
print("T4c: Top 5 Brands")
print(df["Brand"].value_counts().head(5))
print("T4d: Top 5 Cities")
print(df["Registered City"].value_counts().head(5))
print("T4e: Median Price by Fuel Type")

result = df.groupby("Fuel")["Price_lakh"].median()

print(result.sort_values(ascending=False))
print("T4f: Corolla GLI Median Prices")

corolla = df[
    df["Model"].str.contains("Corolla GLI", case=False, na=False) &
    df["Year"].between(2010, 2018)
]

median_prices = corolla.groupby("Year")["Price_lakh"].median()

print(median_prices)

print("Yearly Price Difference:")
print(median_prices.diff())

if len(median_prices) > 1:
    yearly_loss = (
        median_prices.iloc[0] - median_prices.iloc[-1]
    ) / (median_prices.index[-1] - median_prices.index[0])

    print("Average yearly decrease in lakh:", yearly_loss)
    print("T4g: Five Most Expensive Ads")

display(
    df.sort_values(
        by="Price",
        ascending=False
    )[["Brand", "Model", "Year", "Price"]].head(5)
)
print("T4h: Cash vs Installment/Leasing")

cash = df[
    df["Transaction Type"].str.contains(
        "Cash", case=False, na=False
    )
]

installment = df[
    df["Transaction Type"].str.contains(
        "Installment|Leasing", case=False, na=False
    )
]

print("Cash Ads:", len(cash))
print("Cash Median Price:", cash["Price_lakh"].median())

print("Installment/Leasing Ads:", len(installment))
print("Installment/Leasing Median Price:", installment["Price_lakh"].median())

```

```python
The median prices of Corolla GLI cars from 2010 to 2018 were calculated. By comparing the prices of different model years, we can observe how the value 
of a car changes over time. Older cars generally have lower prices than newermodels.
The five most expensive car advertisements were identified by sorting the Price column in descending order. 
These records were checked to identify any unusually high prices that might indicate incorrect data.
Cash and Installment/Leasing advertisements were compared by calculating their total counts and median prices.
This comparison helps us understand the difference between the two transaction types in the used-car market.
```
